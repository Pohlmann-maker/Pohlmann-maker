```python
%%writefile filamente.c
#include <stdio.h>

int main()
{
    float dados[7];

    int impressora;
    int bandeira;

    float potencia;
    float custoFilamento;
    float consumoEnergia;
    float custoEnergia;
    float custoTotal;
    float precoVenda;
    float lucro;

    printf("Digite o preco do filamento por kg: ");
    scanf("%f", &dados[0]);

    printf("Digite o filamento utilizado em gramas: ");
    scanf("%f", &dados[1]);

    printf("Digite a impressora (1-A1, 2-Ender 3, 3-X1C): ");
    scanf("%d", &impressora);

    if (impressora == 1)
    {
        potencia = 200;
    }
    else if (impressora == 2)
    {
        potencia = 120;
    }
    else if (impressora == 3)
    {
        potencia = 350;
    }
    else
    {
        printf("Impressora invalida.\n");
        return 0;
    }

    printf("Digite o tempo de impressao em horas: ");
    scanf("%f", &dados[2]);

    printf("Digite a bandeira (1-Verde, 2-Amarela, 3-Vermelha): ");
    scanf("%d", &bandeira);

    if (bandeira == 1)
    {
        dados[3] = 0.65;
    }
    else if (bandeira == 2)
    {
        dados[3] = 0.80;
    }
    else if (bandeira == 3)
    {
        dados[3] = 1.00;
    }
    else
    {
        printf("Bandeira invalida.\n");
        return 0;
    }

    printf("Digite o custo da mao de obra: ");
    scanf("%f", &dados[4]);

    printf("Digite o custo fixo: ");
    scanf("%f", &dados[5]);

    printf("Digite a margem de lucro em porcentagem: ");
    scanf("%f", &dados[6]);

    custoFilamento =
        (dados[1] / 1000) * dados[0];

    consumoEnergia =
        (potencia / 1000) * dados[2];

    custoEnergia =
        consumoEnergia * dados[3];

    custoTotal =
        custoFilamento +
        custoEnergia +
        dados[4] +
        dados[5];

    precoVenda =
        custoTotal +
        (custoTotal * dados[6] / 100);

    lucro =
        precoVenda - custoTotal;

    printf("\n========================================\n");
    printf("          RESULTADO DO ORCAMENTO\n");
    printf("========================================\n");

    printf("\nPotencia da impressora: %.0f W", potencia);
    printf("\nFilamento utilizado: %.2f g", dados[1]);
    printf("\nCusto do filamento: R$ %.2f", custoFilamento);
    printf("\nConsumo de energia: %.3f kWh", consumoEnergia);
    printf("\nValor do kWh: R$ %.2f", dados[3]);
    printf("\nCusto da energia: R$ %.2f", custoEnergia);
    printf("\nMao de obra: R$ %.2f", dados[4]);
    printf("\nCusto fixo: R$ %.2f", dados[5]);

    printf("\n----------------------------------------");

    printf("\nCUSTO TOTAL: R$ %.2f", custoTotal);
    printf("\nMargem de lucro: %.2f%%", dados[6]);
    printf("\nPRECO DE VENDA: R$ %.2f", precoVenda);
    printf("\nLUCRO: R$ %.2f", lucro);

    printf("\n========================================\n");

    return 0;
}
```
